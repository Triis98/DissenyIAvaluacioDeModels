# Optimització de models supervisats

Ja hem vist moltes millores als models supervisat i els algorismes més importants desenvolupats. Ara es tracta de millorar l'anàlisi per trobar els millors algorismes o hiperparàmetres.

Per exemple, en una regressió polinòmica s'ha de fixar el grau del polinomi amb un hiperparàmetre. Es pot anar provant el millor, però eixes proves es poden automatitzar. 

Molts models tenen hiperparàmetres, els valors dels quals cal fixar abans de començar el procés de
aprenentatge. Alguns com la regressió lineal no tenen hiperparàmetres perquè el resultat és el millor possible a un cost computacional molt gran per a moltes variables o moltes dades. Els algorismes més eficients com els basats en els descens de gradient tenen diferents resultats ens funció de criteris de parada d'entrenament o velocitat d'entrenament. 




## PipeLine

El procés d'entrenament manualment suposa tasques repetitives que poden donar lloc a errades. Si amés s'ha de fer una normalització o selecció de variables, per exemple, cada vegada, ens podem oblidar d'algun pas i el codi queda farragós. Es poden crear `Pipelines` amb totes les tasques ordenades com una `pipe` del paradigma de programació funcional i tractar eixe `pipeline` com un algorisme atòmic complet. 

Ací tenim un exemple amb una `Pipe` per a escalar, seleccionar el millor k i aplicar un model de machine learning. Amés, el pipeline se passa com a paràmetre a un GridSearchCV com un model qualsevol. En aquest capítol estan explicades aquestes tècniques:


```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Ridge
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.preprocessing import StandardScaler


from sklearn.datasets import load_diabetes
import pandas as pd

data = load_diabetes(as_frame=True)
X = data.data
y = data.target   # progresión de la enfermedad (variable continua)

print("Muestras:", X.shape[0])
print("Variables:", X.shape[1])

# Pipeline con ANOVA + modelo de regresión
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("anova", SelectKBest(score_func=f_regression)),
    ("model", Ridge())
])

# Valores candidatos de k
param_grid = {
    "anova__k": [2, 3, 5, 7, 10]
}

# GridSearch con validación cruzada
grid = GridSearchCV(
    pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="r2"
)

grid.fit(X, y)

print("Mejor k:", grid.best_params_["anova__k"])
print("Mejor R² (CV):", grid.best_score_)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

results = pd.DataFrame(grid.cv_results_)

plt.figure(figsize=(6, 4))
plt.plot(results["param_anova__k"], results["mean_test_score"], marker="o")
plt.xlabel("k (nº de variables seleccionadas)")
plt.ylabel("R² medio (CV)")
plt.title("Rendimiento del modelo en función de k")
plt.tight_layout()
plt.show()
```



## Hiperparàmetres i sobreajustament

Se'n mostra un exemple amb un model de regressió polinòmic. El valor de la variable target,
i es calcula amb un polinomi d'una variable predictora x. El grau del polinomi és un hiperparàmetre del
model. Amb valor 1 per a aquest hiperparàmetre, els únics models que es poden obtenir com a resultat del
procés d'entrenament són models lineals de la forma y = α0 + α1x. Si es fixa un valor de 2, es poden
obtenir models en què el valor de i es calcula amb un polinomi de segon grau, y = α0 + α1x + α2x2.

![ajust i sobreajust](./imgsclassificacio/ajust.png)

Una adequada selecció dels hiperparàmetres és determinant per obtenir un model apropiat per
un dataset. Els valors fixats per als hiperparàmetres restringeixen el conjunt de possibles models que es
poden obtenir com a resultat del procés daprenentatge i, per tant, la capacitat dobtenir un model
que s'ajuste a un dataset donat.

El primer dels tres models models anteriors pateix un ajustament insuficient a les dades (underfitting). Perquè un model de regressió lineal és massa senzill per a aquest dataset. L'últim és el que me
jor sajusta a les dades dentrenament, però segurament massa. Hi ha poques dades a la dreta, per
valors alts de la variable predictora, però tenen molta influència en la forma de la corba. Hi ha sobreajustament u overfitting, i aquest model no donarà tan bones prediccions per a dades noves com el segon. És necessàri, per tant, un compromís entre un bon ajustament a les dades d'entrenament i una generalitat suficient.

Tots els algorismes tant de classificació com de regressió que tinguen hiperparàmetres poden patir de sobreajust. En els exemples dels capítols inicials sols hem calculat la precisió com R² o la tassa d'encerts, però ara mirarem mètriques que avaluen si un model està sobreajustat o si és adeqüat per al problema com la precisió i el recall o matrius de confusió. 

## Preparació del dataset

Per evitar Overfitting o un entrenament que no funciona cal plantejar si cal fer alguns d'aquests pasos per preparar el dataset:

**Separació en Train, Test**: Abans de tot cal separar per evitar que les següents tècniques provoquen data leakage del train cap al test.

**Detecció de valors nuls**: En cas de detecció es poden eliminar files o columnes o imputar dades. 

**Detecció de variables molt correlacionades**: Les relacions molt colineals són problemàtiques, es pot intentar eliminiar variables, reduir dimensions o aplicar regularitzacions. 

**Detecció d'outliers de distints tipus**: Un outlier pot ser un valor fora d'escala o soroll, però també pot ser una combinació de valors (multivariats) o contextuals. En eixe cas cal decidir si els llevem, imputem o els deixem perquè són importants per a l'entrenament o són l'objectiu. 

**Codificació de variables categòriques** Amb `Label Encoding` o `One-hot` per a tenir dades numèriques per als models.

**Escalat de les dades**: Posar totes les variables numèriques en una mateixa escala (per exemple, amb StandardScaler o MinMaxScaler). Això és vital per a models lineals, regularitzats (com els Ridge/Lasso), SVMs o xarxes neuronals, perquè si una variable té valors molt alts (com el sou) eclipsarà les altres (com l'edat).



### Featuring Engineering  

Ja hem fet aquesta tècnica en la regressió polinòmica. A partir d'una variable es calculen potències i es creen noves variables que ja es poden entrenar en un model lineal. Però moltes vegades podem crear noves variables a partir d'existents que expliquem millor el problema per als models de machine learning. 

La codificació de variables o escalat també són tècniques dins del `Featuring Engineering`, però amés podem afegir:

* **Extracció de característiques (*Feature Extraction*)**
Consisteix a descompondre una dada complexa per crear múltiples variables més útils.
* **Exemple:** A partir d'una columna de data i hora (`"2026-10-12 14:30"`), crear noves variables independents com `any`, `mes`, `dia_de_la_setmana` i `hora`.  En àmbits complexos sovint se'ls coneix específicament com a característiques de domini (domain-specific features) perquè requereixen conèixer a fons les regles del joc: Dada crua: Un `historial` llarg de transaccions amb el preu i la data de cada compra. Característica nova creada: Les variables del model `RFM` (Recència, Freqüència i Monetització), com ara el `temps mitjà` en dies que passa entre una compra i la següent o el `valor mitjà` del carretó en comparació amb la mitjana de la seva categoria.

* **Discretització o Agrupació en intervals (*Binning*)**
Transforma una variable contínua en categories o trams discrets per reduir el soroll o capturar patrons no lineals.
* **Exemple:** Convertir l'edat exacta d'una persona en categories com `"Jove"` (18-30), `"Adult"` (31-60) i `"Sènior"` (61+).

* **Transformacions matemàtiques**
S'apliquen funcions per corregir distribucions molt esbiaixades (amb valors extrems).
* **Exemple:** Aplicar el logaritme natural a una variable d'ingressos econòmics per suavitzar les diferències extremes i acostar-la a una distribució normal.

* **Imputació de valors nuls**
Omplir o estimar les dades que falten al conjunt de dades original.
* **Exemple:** Substituir els valors buits d'una columna de preus d'habitatges per la mediana de tota la columna.

> Moltes d'aquestes tècniques entren dins del terreny del mòdul de neteja de dades, però cal no oblidar-les. 



### Solucions a Datasets desbalancejats 

En el cas de la classificació, podem trobar-nos en datasets els quals la variable target no té la mateixa quantitat de tots els seus possibles valors. En eixe cas, el valor predominant tindrà més pes a l'entrenament. 

Per solucionar eixe problema es poden utilitzar paràmetres a l'entrenament, llevar files aleatories de la classe majoritària o crear dades sintètiques de la classe minoritària. 

#### class_weight

El paràmetre `class_weight='balanced'` és una eina automàtica de Scikit-Learn que s'utilitza en models com la regressió logística per combatre el desbalanceig de classes. 

Durant l'entrenament, el model calcula de manera autònoma un pes inversament proporcional a la freqüència de cada classe, assignant un factor multiplicador major a la classe minoritària. Això evita que l'algoritme s'esbiaixi cap a la classe predominant i s'aplica de forma homogènia a totes les mostres que pertanyen al mateix grup, oferint una solució sense modificar directament el conjunt de dades original.

#### sample_weight

El paràmetre `sample_weight` del mètode `.fit()` ofereix una opció manual i personalitzable per assignar un pes específic a cada fila de manera individual durant l'entrenament del model. 

A diferència d'enfocaments completament automàtics, aquest paràmetre permet que cada mostra tinga un valor de ponderació únic, la qual cosa resulta útil no només per corregir el desbalanceig de classes mitjançant càlculs a mida, sinó també per prioritzar dades més recents, fiables o d'alt interès estratègic, adaptant així l'aprenentatge a necessitats complexes i específiques del projecte.

### Corves de validació

Als capítols anteriors hem vist una manera estàtica de preparar els models, entrenar i treure mètriques per validar. Aquesta metodologia s'ha de sofisticar un poc més si volem provar distints models o hiperparàmetres.

Una de les maneres de veure quin és l'efecte de modificar un hiperparàmetre i trobar el valor òptim són les corbes de validació.

Veiem un exemple ràpid de regressió polinòmica:

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
%matplotlib inline
import matplotlib.pyplot as plt
from sklearn.model_selection import validation_curve

def PolynomialRegression(degree=2, **kwargs):
    return make_pipeline(PolynomialFeatures(degree),
                         LinearRegression(**kwargs))

def make_data(N, err=1.0, rseed=42):
    # randomly sample the data
    rng = np.random.RandomState(rseed)
    X = rng.rand(N, 1) ** 2
    y = 10 - 1. / (X.ravel() + 0.1)
    if err > 0:
        y += err * rng.randn(N)
    return X, y

X, y = make_data(25, err=1.1)

X_test = np.linspace(-0.1, 1.1, 500)[:, None]

plt.scatter(X.ravel(), y, color='black')
axis = plt.axis()
for degree in [1, 3, 5]:
    y_test = PolynomialRegression(degree).fit(X, y).predict(X_test)
    plt.plot(X_test.ravel(), y_test, label='degree={0}'.format(degree))
plt.xlim(-0.1, 1.0)
plt.ylim(-2, 12)
plt.legend(loc='best')

```

Es pot veure que el grau 3 funciona raonablement bé i el 5 té una forma molt exagerada. Però dibuixar tots els graus no és possible i menys quan estem en problemes de més dimensions. 

Amb `validation_curve` tenim una manera de analitzar el rendiment d'un model amb varis possibles valors d'un paràmetre en concret. 

En aquest cas es validen de 0 a 21 graus en un cross-validation de 7. A pesar de que la regressió lineal actual de `Scikit-learn` és molt robusta, en certes combinacions de quantitat de mostres i random la curva baixa molt conforme augmenten els graus. 

Si provem a augmentar el dataset, la corva s'estabilitza. Però és molt més complex un model de grau 20 que un de grau 2 i cal valorar si mereix la pena o si no tindrà overfitting per a dades més grans o menudes que les de validació actuals. 

La navalla d'Ockham: Davant de dos models amb un rendiment similar, sempre s'ha de triar el més senzill. Un model de grau 2 és molt més estable, fàcil d'interpretar i menys propens a comportaments estranys que un de grau 20. 

> En qualsevol cas els models regularitzats eviten en gran mesura l'overfitting. Per trobar el punt d'equilibri perfecte d'aquesta força de regularització (normalment anomenada alfa o $\alpha$), s'utilitza exactament la mateixa tècnica que hem vist abans: les corbes de validació, però variant l'alfa en comptes del grau del polinomi

```python
degree = np.arange(0, 21)
train_score, val_score = validation_curve(PolynomialRegression(), X, y, param_name='polynomialfeatures__degree', param_range=degree, cv=7)

plt.plot(degree, np.median(train_score, 1), color='blue', label='training score')
plt.plot(degree, np.median(val_score, 1), color='red', label='validation score')
plt.legend(loc='best')
plt.ylim(0, 1)
plt.xlabel('degree')
plt.ylabel('score')

print(np.median(train_score, 1), np.median(val_score, 1))

print("Millor grau:", degree[np.argmax(np.median(val_score, 1))])
```

* **Sobreajust (*overfitting*) amb dades petites:** Si el conjunt de dades és reduït, el model tendeix a memoritzar-lo; la puntuació d'entrenament serà alta, però la de validació serà baixa.
* **Sotajust (*underfitting*) amb dades grans:** A mesura que el dataset creix, el model ho té més difícil per ajustar-se a tot perfectament: la puntuació d'entrenament baixa lleugerament, però la de validació puja perquè generalitza millor.
* **Límit entre corbes:** Llevat de casualitat estadística, el model **mai** obtindrà una puntuació millor al conjunt de validació que al d'entrenament. Per això, les dues corbes s'acosten a mesura que augmenten les dades, però no s'arriben a creuar.

### Validació creuada (Cross Validation)

Limitacions del mostreig anterior:

* Els resultats depenen massa de l'atzar. Si el dataset és gran no és un gran problema.
* Si cal explorar varis hiperparàmetres es complica, si són més d'un fa falta un bucle niuat. 

Per evitar eixos problemes es poden fer varies particions distintes i treue la mitjana dels resultats. Però el problema és que pot ser que alguna mostra no es contemple mai a l'entrenament. 

La validació creuada fa un mostreig més sistematic: Fa k particions aleatories i entrena amb totes les combinacions de train i test agafant una partició de tests distinta cada vegada. D'aquesta manera es prova l'entrenament i el test amb totes les dades. El resultat és la mitjana dels resultats. 



## GridSearchCV

L'experiència amb models i dades ens ajudarà a intuir el millor model per a cada cas. Hi ha algunes regles a ull i es pot anar provant de models més senzills a més pesats. Però professionalment el que cal fer és sistematitzar la búsqueda dels millors models i dels millors paràmetres. 

En Scikit-Learn tenim `GridSearchCV` per a seleccionar els hiperparàmetres o models óptims. 

Permet indicar tots els possibles valors i combinacions de varis paràmetres i fer una validació creuada amb tots ells. Es diu búsqueda en quadrícula. 

Altres possibilitats són 

Per fer una cerca aleatòria o amb optimització en lloc d'una cerca exhaustiva en quadrícula (*Grid Search*), `scikit-learn` i l'ecosistema de Python disposen d'eines molt potents i fàcils d'utilitzar:

1. Cerca aleatòria: `RandomizedSearchCV`

En lloc de provar totes les combinacions possibles d'una quadrícula (cosa que pot ser molt lenta si tens molts hiperparàmetres), **`RandomizedSearchCV`** selecciona de manera aleatòria un nombre determinat de combinacions (`n_iter`).

Això és molt útil quan tens espais de cerca grans o continus, ja que sovint troba resultats gairebé tan bons com el *Grid Search* invertint una fracció del temps.

2. Optimització Bayesiana / Estocàtica avançada (Optuna o Scikit-Optimize)

Quan parles d'"optimització estocàtica" o intel·ligent (on cada prova depèn dels resultats de les anteriors per buscar el mínim/màxim de forma eficient), `scikit-learn` per si sol no té un mètode natiu integrat més enllà del `Grid` i el `Random`, però s'integra perfectament amb llibreries externes com **Optuna** (la més recomanada avui dia) o `scikit-optimize`.

Amb **Optuna**, l'optimització és "intel·ligent": aprèn de les proves fallides o poc òptimes per descartar zones dolentes i centrar-se en les prometedores de manera iterativa.

* **`GridSearchCV`:** Si l'espai de paràmetres és molt petit i vols provar absolutament totes les combinacions possibles de manera exhaustiva.
* **`RandomizedSearchCV`:** Si tens molts paràmetres o valors continus i vols una cerca ràpida i aleatòria nativa de `scikit-learn`.
* **Optuna (Optimització Bayesiana):** Si vols que l'algorisme sigui "intel·ligent" i aprengui de cada prova per trobar els millors hiperparàmetres amb molta menys estona i major precisió.


## Early Stopping

Molts entrenaments són molt lents. Quan s'està fent un entrenament per descens de gradient, no sempre més iteracions impliquen una millora. De vegades no millora considerablements, de vegades arriba a un mínim local i de vegades inclús empijora perquè fa `Overfitting`. 

Per evitar eixos problemes o poder fer més proves en el mateix temps es pot implementar un entrenament oer etapes on s'avalua el model cada certs dades o iteracions i es decidexi si encara està millorant, si el model és suficientment bó o no val la pena seguir entrenant amb eixos hiperparàmetres.

### Selecció de variables

Identificant les variables que expliquen o serveixen per predir l'objectiu i eliminant les que són irrellevants podem reduir dimensions de forma senzilla i millorem l'explicabilitat dels resultats.

Els models basat en arbres ja tenen la seua manera de detectar les variables importants `MDI / Mean Decrease in Impurity`. En models lineals també es fa una selecció automàtica en els regularitzats. Inclús en un model lineal no regularitzat es pot veure el valor absolut dels coeficients. 

Per fer un anàlisi més detallat podem recurrir a diverses tècniques: 

#### Permutation Importance

https://scikit-learn.org/stable/modules/generated/sklearn.inspection.permutation_importance.html#sklearn.inspection.permutation_importance


Funciona mesurant quant cau el rendiment del model (per exemple, l'exactitud o l'error quadràtic mig) quan barregem de manera aleatòria els valors d'una sola variable, trencant la seva relació amb la resta.

Funciona amb qualsevol tipus de model, és lenta i no funciona del tot bé si hi ha variables altament correlacionades, ja que una pot suplir a l'altra. 


#### Partial Dependence

Si la Permutation Importance diu quants de "pes" o importància té una variable, el PDP diu exactament quin tipus de relació (la forma) té aquesta variable amb la predicció.

Un gràfic de dependència parcial mostra l'efecte marginal d'una o dues variables sobre la predicció d'un model de Machine Learning.

En relacions no lineals algunes variables són importants sols en un rang de valors. Per exemple: "A partir de 150 metres quadrats, el preu del pis deixa de pujar de manera lineal i s'estanca".

### ANOVA

El mètode **ANOVA** va un pas més enllà que les correlacions `Pearson` o `Spearman` explicant si una variable explica una alta. 

Aquest mètode ajusta un model senzill entre cada variable i el `target`. Separa la variabilitat total que és explicada de la variable objectiu. I avalua si la variabilitat és major que l'atzar. 

`ANOVA` funciona tant per a classificació com per a regressió. 

Es tracta d'un pas previ a entrenar un model i permet descartar variables o inclús fer un informe d'un dataset sense necessitat de fer un model. Forma part del procés EDA. 

#### ANOVA per a Regressió



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import SelectKBest, f_classif, RFE
from sklearn.ensemble import RandomForestClassifier
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

import shap
shap.initjs()


from sklearn.datasets import load_diabetes
import pandas as pd

data = load_diabetes(as_frame=True)
X = data.data
y = data.target   # progresión de la enfermedad (variable continua)

print("Muestras:", X.shape[0])
print("Variables:", X.shape[1])

```

```python
# Escalado de las variables
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=X.columns,
    index=X.index
)

X_scaled_df.head()
```

Anem a crear un dataset amb noves variables de pur soroll. 

```python
import numpy as np
import pandas as pd

# Semilla para reproducibilidad
np.random.seed(42)

# Filtrar variables numéricas (Diabetes ya lo es, pero se deja como plantilla)
X_num = X.select_dtypes(include=["number"])

# Número de variables de ruido
n_noise = 40

# Generación de variables irrelevantes
X_noise = pd.DataFrame(
    np.random.randn(len(X_num), n_noise),
    columns=[f"noise_{i}" for i in range(n_noise)],
    index=X_num.index
)

# Dataset aumentado con ruido
X_noisy = pd.concat([X_num, X_noise], axis=1)

X_noisy.shape
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.model_selection import cross_val_score

baseline_pipeline_noisy = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", Ridge())
])

baseline_noisy_scores = cross_val_score(
    baseline_pipeline_noisy,
    X_noisy,
    y,
    cv=5,
    scoring="r2",
    error_score="raise"
)

print("R² medio (sin ANOVA, con ruido):", baseline_noisy_scores.mean())
print("Desviación estándar:", baseline_noisy_scores.std())
```

```python
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.model_selection import GridSearchCV

pipeline_tmp_noisy = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("anova", SelectKBest(score_func=f_regression)),
    ("model", Ridge())
])

param_grid = {
    "anova__k": [5, 10, 20, 30, 40]
}

grid_noisy = GridSearchCV(
    pipeline_tmp_noisy,
    param_grid=param_grid,
    cv=5,
    scoring="r2",
    error_score="raise"
)

grid_noisy.fit(X_noisy, y)

grid_noisy.best_params_, grid_noisy.best_score_


print("R² sin ANOVA (con ruido):", baseline_noisy_scores.mean())
print("R² con ANOVA (con ruido):", grid_noisy.best_score_)
print("Mejor k:", grid_noisy.best_params_["anova__k"])

import pandas as pd
import matplotlib.pyplot as plt

results = pd.DataFrame(grid_noisy.cv_results_)

plt.figure(figsize=(6, 4))
plt.plot(results["param_anova__k"], results["mean_test_score"], marker="o")
plt.xlabel("k (nº de variables seleccionadas)")
plt.ylabel("R² medio (CV)")
plt.title("Efecto de ANOVA con variables de ruido")
plt.tight_layout()
plt.show()

```

El F score permet ordenar les variables per la seua capacitat explicativa individual:

```python
# Extraer el pipeline óptimo
best_pipeline = grid_noisy.best_estimator_

# Extraer el selector ANOVA entrenado
selector_best = best_pipeline.named_steps["anova"]

# Variables seleccionadas
selected_features = X_noisy.columns[selector_best.get_support()]

# Ranking ANOVA (coherente con X_noisy)
anova_scores = pd.Series(
    selector_best.scores_,
    index=X_noisy.columns
).sort_values(ascending=False)

# Visualización (resaltando las seleccionadas)
selected = set(selected_features)

colors = [
    "tab:orange" if feature in selected else "tab:blue"
    for feature in anova_scores.head(10).index
]

plt.figure(figsize=(8, 4))
anova_scores.head(10).plot(kind="bar", color=colors)
plt.ylabel("F-score (ANOVA - f_regression)")
plt.title("Ranking ANOVA tras GridSearch (con ruido)")
plt.tight_layout()
plt.show()
```

La validació creauada ha permes seleccionar el k de variables que maximitza la regressió. 

És el punt de compromís entre la complexitat i la seua capacitat.

El `summary_plot` mostra la importància global de cada variable el sentit positiu o negatiu i la distribució dels efectes. 

En aquest exemple es veu que:

* L'eix horizontal indica l'impacte de cada variable sobre la predicció. Els valors alts incrementen la predicció i els baixos la redueixen. Per tant bmi i s5 amb valors alts són molt influyents en la predicció positiva i els valors baixos impacten en la reducció de la predicció.

El model basa les seues prediccións en l'estat del pacient. La variable que més influencia té és l'index de massa corporal o `bmi`. Els indicadors metabólics (s5) també són molt rellevants. Per tant podem concloure que `bmi` és una variable clau i el seu efecte en la predicció és extable, monotónic i molt lineal. Aquest gràfic permet detectar relacions no lineals que no siguen evidents.

Exemple complet:

```python
import shap
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
import pandas as pd

# 1) Cargar datos
data = load_diabetes(as_frame=True)
X = data.data
y = data.target

# 2) Pipeline simple (lineal, interpretable)
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", Ridge(alpha=1.0))
])

pipeline.fit(X, y)

# 3) SHAP
X_scaled = pipeline.named_steps["scaler"].transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

explainer = shap.Explainer(pipeline.named_steps["model"], X_scaled_df)
shap_values = explainer(X_scaled_df)

# 4) Buscar una observación con contribuciones positivas y negativas
i = np.where(
    (shap_values.values > 0).any(axis=1) &
    (shap_values.values < 0).any(axis=1)
)[0][0]

# 5) Force plot
shap.initjs()
shap.plots.force(
    shap_values.base_values[i],
    shap_values.values[i],
    X.iloc[i]
)
```
