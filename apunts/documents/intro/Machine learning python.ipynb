També es pot sustuir per d'una manera matemàticament més precisa amb `SimpleImputer`.

El comportament del `SimpleImputer` es defineix principalment amb el paràmetre `strategy`, que pot prendre els valors següents:

* **`'mean'` (Mitjana):** Substitueix els valors faltants per la **mitjana** de la columna. Només funciona per a columnes numèriques. És sensible a outliers.
* **`'median'` (Mediana):** Substitueix els buits per la **mediana** de la columna. Només per a dades numèriques. És molt robusta davant d'outliers.
* **`'most_frequent'` (Moda):** Substitueix els valors faltants pel **valor més freqüent** (la moda) de la columna. Molt útil tant per a variables numèriques com **categòriques**.
* **`'constant'` (Constant):** Omple els buits amb un valor fix definit per tu mitjançant el paràmetre `fill_value` (per exemple, `0` per a números o `"Desconegut"` per a textos).



````python
import numpy as np
from sklearn.impute import SimpleImputer

# Dades amb valors faltants (np.nan)
X = [[1, 2], [np.nan, 3], [7, 6], [4, np.nan]]

# Definim l'imputer perquè utilitzi la mitjana de cada columna
imputer = SimpleImputer(strategy='mean')

# Ajustem i transformem les dades
X_imputed = imputer.fit_transform(X)

print(X_imputed)
````

Inclús es pot fer un menut algorisme de `machine learning` per imputar a partir de les dades de registres semblants. Per a això tenim `KNNImputer`. El problema és que és molt lent amb moltes dades i les necessita estandaritzades. 

````python
import numpy as np
from sklearn.impute import KNNImputer

# Dades amb valors faltants
X = [[1, 2, np.nan], 
     [3, 4, 3], 
     [np.nan, 6, 5], 
     [8, 8, 7]]

# Definim el KNNImputer amb k=2 veïns
imputer = KNNImputer(n_neighbors=2)

X_imputed = imputer.fit_transform(X)
print(X_imputed)
````

### Normalització i estandardització

Les variables numèriques poden tenir rangs molt diversos i extrems. Molts models no funcionen bé amb variables de distints rangs, ja que li donen més influència a unes variables que a altres artificialment. 

Per això és important normalitzar les dades de manera que tinguen una escala similar. 

Les opcions són:

* Normalització: Fer que totes les dades vagen de 0 a 1 o de -1 a 1 (`MinMaxScaler`) 
* Estandarització: fer que les dades tinguen variança 1 i mitjana 0. No tenen límits fixos, però anirà d'un valor negatiu a un positiu. (`StandardScaler`) Aquesta és l'elecció per defecte. 

La normalització o funciona gens amb outliers (valors atípics). Eixe és el punt feble principal de qualsevol normalització basada en el mínim i el màxim (com el Min-Max de 0 a 1 o de -1 a 1).
Aquesta normalització depèn exclusivament i de manera estricta del valor mínim i del valor màxim absoluts de les teues dades. Un número molt gran pot fer que la resta queden comprimits en un rang minúscul. `StandardScaler` és més robust. Si realment tenim valors atípics que no podem llevar i és important que no afecte, `RobustScaler` utilitza la mediana i ignora, per tant, valors extrems, però trenca l'escala i dona resultats menys eficients si no és necessari. 

> Si coneguem exactament el rang, podem fer un divisió a totes les dades per el màxim i tenim un resultat igual que la normalització i més ràpid. Per exemple en el cas `MNIST` es sol fer al tenir de 0 a 255. 

````python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# 1. Seleccionem només les columnes que són de tipus float64
columnes_float = df.select_dtypes(include=['float64']).columns

# 2. Inicialitzem l'StandardScaler i l'MinMaxScaler
scaler = StandardScaler()
minmax_scaler = MinMaxScaler()

df_minmax = df.copy()

# 3. Apliquem l'escalat a aquestes columnes i les tornem a assignar al DataFrame

df[columnes_float] = scaler.fit_transform(df[columnes_float])
df_minmax[columnes_float] = minmax_scaler.fit_transform(df_minmax[columnes_float])

df.describe()  # Estadístiques descriptives després de l'escalat
#df_minmax.describe()  # Estadístiques descriptives després de l'escalat amb MinMaxScaler
````

### One-hot Encoding

En cas de voler fer un algorisme de regressió, aquest sols accepta atributs numèrics. Els atributs categòrics que són rellevants s'han de codificar com a numèrics. 

Hi ha un problema amb posar números als atributs categòrics. Si fem `Label encoding` que és assignar un número a cada categoria, li donarem un significat diferent al que volem. 

Suposem el cas d'una regressió lineal amb dos variables. L'equació resultant serà: 

En el cas d'una regressió lineal amb dues variables independents (també coneguda com a **regressió lineal múltiple** amb dues variables predictoras), l'equació que defineix la relació entre les variables és la següent:

$$Y = \beta_0 + \beta_1X_1 + \beta_2X_2 + \epsilon$$

---

Desglossament dels components:

* **$Y$**: És la **variable dependent** (o variable de resposta), que és el valor que volem predir o explicar.
* **$X_1$ i $X_2$**: Són les **variables independents** (o variables explicatives/predictoras).
* **$\beta_0$**: És la **intercepció** (o terme constant). Representa el valor esperat de $Y$ quan tant $X_1$ com $X_2$ són iguals a zero.
* **$\beta_1$ i $\beta_2$**: Són els **coeficients de regressió** (o pendents parcials). Indiquen quant canvia $Y$ per cada unitat que augmenta la variable corresponent ($X_1$ o $X_2$), mantenint l'altra variable constant.
* **$\epsilon$**: És el **terme d'error** (o residu). Representa la variabilitat en $Y$ que no pot ser explicada pel model lineal amb les variables donades.

A diferència d'una regressió lineal simple (on la representació és una línia recta en un pla 2D), aquesta equació representa un **pla en un espai tridimensional**.

Si tenim una variable, per exemple, *marca*  en un dataset de cotxes per predir el preu, si són 5 marques tindrem `[0,1,2,3,4]` o `[1,2,3,4,5]` o `[-1, -0.5, 0, 0.5, 1]`. En qualsevol cas tenim números que acabaran afectant a un $\beta$ i multiplicats per un X i el seu valor més alt o baix influirà numèricament encara que siga una variable categòrica que no implica un valor més alt o més baix. 

La solució és fer `One-hot encoding` al crear una columa dicotòmica amb 0 o 1 per cada categoria. D'aquesta manera per a 5 marques es poden fer 5 columnes, una per marca indicant amb un 1 si és eixa marca. El model aprendrà a asignar una  $\beta$ determinada a marques generalment mès valuoses sense afectar a la resta del model. 

Una millor manera és fer `Variables Dummy` llevant una columna. A 5 marques 4 columnes, una per marca i si és 0 la marca que falta (la marca per defecte). Això evita una correlació perfecta entre eixes columnes que, si no, sempre sumen 1 i crea una dependència lineal que empijora el model. 

https://pandas.pydata.org/docs/reference/api/pandas.get_dummies.html

### Correlacions entre atributs

Existeix correlació entre dos atributs quan l'augment dels valors d'un conlleva l'augment del
valor de otro (correlación positiva) o la seva disminución (correlación negativa).

Per exemple, hi ha correlació positiva entre l'alçada de les persones i el seu pes. Per terme mitjà, les persones pesen més quant més altes son. Hi ha correlació negativa entre el nombre d'anys d'escolarització de les personas y el número de faltes de ortografía que cometen.
Interesa estudiar correlacions entre atributs, pels següents motius:
* Es pot obtindre un model més sencill eliminant atributs que no proporcionen molta informació respecte a altres. Açò pot suposar una merma en la qualitat del model, que hi haurà que valorar.
* Per a models de regressió, interessa especialment els atributs predictors que tenen major correlació amb l'atribut target u objectiu.
* Per a models de regressió, pot ser perjudicial tenir atributs predictors que siguen totalment correlacionats amb l'objectiu, ja que el model podria donar massa importància a aquest atributs i no tenir en compte els altres. El motiu pot ser que continga implícitament la resposta (`target leakage`) i per tant, les noves dades no tindran aquesta variable. O que tinga `Overfitting` i siga massa depenent d'aquesta variable. En qualsevol cas, una correlació  `1.0` és sospitosa i ens pot indicar un error de disseny.

La correlació que hem vist anteriorment és la lineal o `Pearson`. Però una variable pot estar correlacionada amb altres de moltes maneres. Una de les més importants també és la correlació `Spearman`, la qual no mira si les dos variables formen una línia sino si quan una creix també creix l'altra. Aquesta junt amb la `Kendall` estan directament disponibles a `Pandas`:

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

X =  np.random.rand(100, 1)
y = 3 * X[:, 0]**4 + 1 + np.random.randn(100) * 0.01

df = pd.DataFrame({'X0': X[:, 0], 'y': y})
df.head()
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Creem una figura amb 1 fila i 3 columnes
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Gràfic de dispersió (Scatter plot)
sns.scatterplot(data=df, x='X0', y='y', ax=axes[0])
axes[0].set_title('Dispersió: X0 vs y')
axes[0].set_xlabel('X0')
axes[0].set_ylabel('y')

# 2. Mapa de calor de Spearman
correlacions_spearman = df.corr(method='spearman')
sns.heatmap(correlacions_spearman, square=True, annot=True, vmin=-1, vmax=1, cmap='coolwarm', ax=axes[1])
axes[1].set_title('Correlació Spearman')

# 3. Mapa de calor de Pearson
correlacions_pearson = df.corr(method='pearson')
sns.heatmap(correlacions_pearson, square=True, annot=True, vmin=-1, vmax=1, cmap='coolwarm', ax=axes[2])
axes[2].set_title('Correlació Pearson')

# Ajustem el disseny perquè no se superposin i mostrem la figura
plt.tight_layout()
plt.show()
````

Un exemple complet i més complex al que creem un dataset de 3 tipus de correlacions i les representem juntes a un mateix gràfic.

````python

n = 1000
np.random.seed(42)
y = np.linspace(-5, 5, n)

x_pearson = 2 * y + np.random.normal(0, 0.5, n)
x_spearman = np.exp(0.3 * y) + np.random.normal(0, 0.2, n)
x_mi = (y ** 2) + np.random.normal(0, 0.2, n)

df = pd.DataFrame(
    {
        "y": y,
        "x_pearson": x_pearson,
        "x_spearman": x_spearman,
        "x_mi": x_mi,
    }
)

# 2. Transformem el DataFrame a format llarg (tidy) per a Seaborn
df_melted = df.melt(
    id_vars=["y"],
    value_vars=["x_pearson", "x_spearman", "x_mi"],
    var_name="Tipus_Relacio",
    value_name="x_valor",
)

# 3. Creem el gràfic amb Seaborn (utilitzant FacetGrid / relplot)
sns.set_theme(style="whitegrid")

g = sns.relplot(
    data=df_melted,
    x="x_valor",
    y="y",
    hue="Tipus_Relacio",  # Assigna un color diferent a cada variable
    col="Tipus_Relacio",  # Separa-ho en 3 subgràfics (facetes)
    kind="scatter",
    alpha=0.6,
    palette="deep",
    facet_kws={"sharex": False},  # Permet que cada eix X s'adapti a la seva escala
)

# Personalitzem els títols i etiquetes
g.set_axis_labels("Valor de X", "Variable Objectiu (y)")
g.set_titles(col_template="{col_name}")

plt.suptitle(
    "Comparació de relacions: Lineal, Monòtona i Parabòlica", y=1.05, fontsize=14
)
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris

iris = load_iris()
X = iris.data
y = iris.target

# Separar les dades en un conjunt d'entrenament i un de prova (80% entrenament, 20% prova)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0,
                                  train_size=0.8)


print("Conjunt d'entrenament:", X_train.shape, y_train.shape)
print("Conjunt de prova:", X_test.shape, y_test.shape)
````
