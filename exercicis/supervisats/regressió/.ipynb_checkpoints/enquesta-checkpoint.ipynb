```python
import pandas as pd
import numpy as np



df = pd.read_csv("salida_traducida.csv", sep=";", encoding="utf-8")
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
df_str = df.select_dtypes(include=['object', 'string'])
```

```python
df_str.describe()
```

```python

df_str["Número de menores de 18 años en el hogar"] = df_str["Número de menores de 18 años en el hogar"]
.replace("No, no puedo.", 0)

df_str["Número de menores de 18 años en el hogar"] = df_str["Número de menores de 18 años en el hogar"]
.replace("Sí, número de tiempo completo: 2", 2)
df_str["Número de menores de 18 años en el hogar"] = df_str["Número de menores de 18 años en el hogar"]
.replace("Sí, número de tiempo completo: 1", 1)
df_str["Número de menores de 18 años en el hogar"] = df_str["Número de menores de 18 años en el hogar"]
.replace("Sí, número de tiempo completo:3", 3)
df_str["Número de menores de 18 años en el hogar"] = df_str["Número de menores de 18 años en el hogar"]
.replace("Sí, número de tiempo parcial: 1", 0.5)
df_str["Número de menores de 18 años en el hogar"] = df_str["Número de menores de 18 años en el hogar"]
.replace("Sí, número de tiempo parcial: 2", 1)
df_str["Número de menores de 18 años en el hogar"].value_counts()
```

```python
# Crear un DataFrame resumen para las 81 columnas de texto
resumen_str = pd.DataFrame({
    'No Nulos': df_str.notnull().sum(),
    'Nulos': df_str.isnull().sum(),
    '% Nulos': (df_str.isnull().mean() * 100).round(2),
    'Valores Únicos': df_str.nunique(),
    'Tipo de Dato': df_str.dtypes
})

# Mostrar el resumen ordenado por la cantidad de valores nulos
print(resumen_str.sort_values(by='Nulos', ascending=False))
```
